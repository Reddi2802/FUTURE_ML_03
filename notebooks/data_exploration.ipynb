```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
from sentence_transformers import util

resume_text = resume_df["Resume_str"].iloc[0]

job_description = """
Cybersecurity Analyst

Requirements:
- Risk assessment
- Vulnerability management
- Security compliance
- Information assurance
- System security
"""
```

```python
resume_embedding = model.encode(
    resume_text,
    convert_to_tensor=True
)

job_embedding = model.encode(
    job_description,
    convert_to_tensor=True
)

similarity = util.cos_sim(
    resume_embedding,
    job_embedding
)

print(float(similarity))
```

```python
cyber_resume = """
Information Assurance Professional
Vulnerability Management
Risk Assessment
Security Accreditation
Security Compliance
Information Security
"""

cyber_embedding = model.encode(
    cyber_resume,
    convert_to_tensor=True
)

job_embedding = model.encode(
    job_description,
    convert_to_tensor=True
)

similarity = util.cos_sim(
    cyber_embedding,
    job_embedding
)

print(float(similarity))
```

```python
it_resume = resume_df[
    resume_df["Category"] == "INFORMATION-TECHNOLOGY"
]["Resume_str"].iloc[0]

hr_resume = resume_df[
    resume_df["Category"] == "HR"
]["Resume_str"].iloc[0]

bd_resume = resume_df[
    resume_df["Category"] == "BUSINESS-DEVELOPMENT"
]["Resume_str"].iloc[0]
```

```python
cyber_jd = """
Cybersecurity Analyst

Requirements:
- Vulnerability Management
- Risk Assessment
- Security Compliance
- Information Assurance
- Security Operations
"""
```

```python
it_emb = model.encode(it_resume, convert_to_tensor=True)
hr_emb = model.encode(hr_resume, convert_to_tensor=True)
bd_emb = model.encode(bd_resume, convert_to_tensor=True)

jd_emb = model.encode(cyber_jd, convert_to_tensor=True)
```

```python
print("IT:", float(util.cos_sim(it_emb, jd_emb)))
print("HR:", float(util.cos_sim(hr_emb, jd_emb)))
print("Business:", float(util.cos_sim(bd_emb, jd_emb)))
```
